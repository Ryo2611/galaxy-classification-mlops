# Error Analysis

Week 2 error analysis for Galaxy Zoo morphology prediction. This notebook inspects class-wise error, high-error examples, and systematic over/under-prediction.

```python
from pathlib import Path
import sys
import json
import matplotlib.pyplot as plt
import pandas as pd

ROOT = Path.cwd()
if not (ROOT / 'src').exists():
    ROOT = ROOT.parent
if str(ROOT) not in sys.path:
    sys.path.insert(0, str(ROOT))

from src.training.error_analysis import summarize_errors

LABELS = ROOT / 'data/external/training_solutions_rev1.csv'
PREDICTIONS = ROOT / 'reports/resnet18_predictions.csv'
OUTPUT = ROOT / 'reports/error_analysis_summary.json'

if PREDICTIONS.exists():
    summary = summarize_errors(str(LABELS), str(PREDICTIONS), str(OUTPUT))
else:
    summary = None
    print('Run src/training/evaluate.py first to create reports/resnet18_predictions.csv')
summary
```

```python
if summary:
    class_rmse = pd.Series(summary['top_classwise_rmse']).sort_values()
    ax = class_rmse.plot(kind='barh', figsize=(8, 5))
    ax.set_xlabel('RMSE')
    ax.set_title('Highest Class-wise RMSE')
    plt.tight_layout()
    plt.savefig(ROOT / 'reports/figures/classwise_rmse_top10.png', dpi=160)
    plt.show()
```
